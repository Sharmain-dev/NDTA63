## 4 )Database Integration

### 4.2 Build the Database (SQLite)
We use **SQLite** via Python's built-in `sqlite3` module. SQLite needs no separate server or password, so the notebook runs identically on any machine (important for the GitHub repository / reproducibility requirement). Two tables are created: `art_coverage` and `under5_mortality`, mirroring the cleaned DataFrames from Section 1.

```python
import sqlite3
import os

DB_PATH = "south_africa_health.db"

# Start from a clean database each run so the notebook is fully reproducible
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

try:
    conn = sqlite3.connect(DB_PATH)
    cursor = conn.cursor()

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS art_coverage (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            country TEXT NOT NULL,
            year INTEGER NOT NULL,
            art_coverage_percent REAL,
            indicator TEXT,
            unit TEXT
        )
    """)

    cursor.execute("""
        CREATE TABLE IF NOT EXISTS under5_mortality (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            country TEXT NOT NULL,
            year INTEGER NOT NULL,
            mortality_per_1000 REAL,
            indicator TEXT,
            unit TEXT
        )
    """)

    conn.commit()
    print("Database created at:", DB_PATH)
    print("Tables created: art_coverage, under5_mortality")

except sqlite3.Error as e:
    print("Database error while creating tables:", e)
```

### 4.3 Load the Cleaned DataFrames into the Database

```python
try:
    # art_sa already covers 2010-2024; mortality_common is the 2010-2024 subset created in Section 1
    art_records = list(
        art_sa[["Country", "Year", "ART_Coverage_Percent", "Indicator", "Unit"]]
        .itertuples(index=False, name=None)
    )
    mortality_records = list(
        mortality_common[["Country", "Year", "Under5_Mortality_per_1000", "Indicator", "Unit"]]
        .itertuples(index=False, name=None)
    )

    cursor.executemany(
        "INSERT INTO art_coverage (country, year, art_coverage_percent, indicator, unit) "
        "VALUES (?, ?, ?, ?, ?)",
        art_records
    )
    cursor.executemany(
        "INSERT INTO under5_mortality (country, year, mortality_per_1000, indicator, unit) "
        "VALUES (?, ?, ?, ?, ?)",
        mortality_records
    )
    conn.commit()

    art_count = cursor.execute("SELECT COUNT(*) FROM art_coverage").fetchone()[0]
    mort_count = cursor.execute("SELECT COUNT(*) FROM under5_mortality").fetchone()[0]

    print(f"Inserted {art_count} rows into art_coverage")
    print(f"Inserted {mort_count} rows into under5_mortality")

except sqlite3.Error as e:
    print("Database error while inserting data:", e)
    conn.rollback()
```

### 4.4 Query the Database
Below are three types of queries: a **JOIN** that combines both tables on `Year`, a **filtered** query, and an **aggregate** query.

```python
try:
    # 1) JOIN query - combine ART coverage and mortality on Year and Country
    join_query = """
        SELECT a.year AS Year,
               a.art_coverage_percent AS ART_Coverage_Percent,
               m.mortality_per_1000 AS Under5_Mortality_per_1000
        FROM art_coverage a
        JOIN under5_mortality m
            ON a.year = m.year AND a.country = m.country
        ORDER BY a.year
    """
    joined_df = pd.read_sql_query(join_query, conn)
    print("JOIN query result (ART coverage & mortality by year):")
    display(joined_df)

    # 2) Filtered query - only recent years
    recent_query = "SELECT * FROM art_coverage WHERE year >= 2020 ORDER BY year"
    recent_df = pd.read_sql_query(recent_query, conn)
    print("\nFiltered query result (ART coverage, 2020 onwards):")
    display(recent_df)

    # 3) Aggregate query - summary statistics computed in SQL
    agg_query = """
        SELECT
            ROUND(AVG(art_coverage_percent), 2) AS avg_art,
            MIN(art_coverage_percent) AS min_art,
            MAX(art_coverage_percent) AS max_art
        FROM art_coverage
    """
    agg_result = cursor.execute(agg_query).fetchone()
    print("\nAggregate query result (ART coverage):")
    print(f"Average: {agg_result[0]}%  |  Min: {agg_result[1]}%  |  Max: {agg_result[2]}%")

except sqlite3.Error as e:
    print("Database error while querying:", e)
```

### 4.5 Update a Record Safely
Parameterised queries (`?` placeholders) are used instead of string formatting to prevent SQL injection and to safely update a single, specific row.

```python
try:
    target_year, target_country = 2024, "South Africa"

    print("Before update:")
    display(pd.read_sql_query(
        "SELECT * FROM art_coverage WHERE year = ? AND country = ?",
        conn, params=(target_year, target_country)
    ))

    # Safely update ART coverage for 2024 (parameterised query - no string concatenation)
    cursor.execute(
        "UPDATE art_coverage SET art_coverage_percent = ? WHERE year = ? AND country = ?",
        (68.0, target_year, target_country)
    )
    conn.commit()
    print(f"\nRows updated: {cursor.rowcount}")

    print("\nAfter update:")
    display(pd.read_sql_query(
        "SELECT * FROM art_coverage WHERE year = ? AND country = ?",
        conn, params=(target_year, target_country)
    ))

    # Revert back to the original value so downstream analysis stays consistent
    cursor.execute(
        "UPDATE art_coverage SET art_coverage_percent = ? WHERE year = ? AND country = ?",
        (67.0, target_year, target_country)
    )
    conn.commit()
    print("\nValue reverted to original (67.0) to keep the dataset consistent.")

except sqlite3.Error as e:
    print("Database error while updating:", e)
    conn.rollback()
```

### 4.6 Delete a Record Safely
To demonstrate a safe delete without losing real data, a temporary test record is inserted and then removed using a parameterised `DELETE` statement.

```python
try:
    # Insert a temporary/test record
    cursor.execute(
        "INSERT INTO art_coverage (country, year, art_coverage_percent, indicator, unit) "
        "VALUES (?, ?, ?, ?, ?)",
        ("TEST_COUNTRY", 1900, 0.0, "Test record", "Test")
    )
    conn.commit()
    before_count = cursor.execute("SELECT COUNT(*) FROM art_coverage").fetchone()[0]
    print(f"Row count before delete: {before_count}")

    # Safely delete only the test record (parameterised WHERE clause)
    cursor.execute(
        "DELETE FROM art_coverage WHERE country = ? AND year = ?",
        ("TEST_COUNTRY", 1900)
    )
    conn.commit()
    print(f"Rows deleted: {cursor.rowcount}")

    after_count = cursor.execute("SELECT COUNT(*) FROM art_coverage").fetchone()[0]
    print(f"Row count after delete: {after_count}")

except sqlite3.Error as e:
    print("Database error while deleting:", e)
    conn.rollback()
```

### 4.7 Load Database Data into Pandas and Export to CSV

```python
try:
    final_query = """
        SELECT a.year AS Year,
               a.art_coverage_percent AS ART_Coverage_Percent,
               m.mortality_per_1000 AS Under5_Mortality_per_1000
        FROM art_coverage a
        JOIN under5_mortality m
            ON a.year = m.year AND a.country = m.country
        ORDER BY a.year
    """
    db_df = pd.read_sql_query(final_query, conn)
    display(db_df)

    db_df.to_csv("sa_health_from_database.csv", index=False)
    print("\nExported query result to 'sa_health_from_database.csv'")

except sqlite3.Error as e:
    print("Database error while exporting:", e)
finally:
    conn.close()
    print("\nDatabase connection closed.")
```

## 5) Python/Excel Data Analysis
This section builds on `db_df` (loaded from the database) by transforming the data further and exporting a formatted Excel workbook with conditional formatting and a native chart.

### 5.1 Clean and Transform the Data

```python
try:
    excel_df = db_df.copy()

    # Year-on-year change for both indicators
    excel_df["ART_YoY_Change"] = excel_df["ART_Coverage_Percent"].diff().round(2)
    excel_df["Mortality_YoY_Change"] = excel_df["Under5_Mortality_per_1000"].diff().round(2)

    # 3-year rolling average to smooth short-term fluctuation in ART coverage
    excel_df["ART_3yr_Rolling_Avg"] = (
        excel_df["ART_Coverage_Percent"].rolling(window=3).mean().round(2)
    )

    # Categorise ART coverage into bands for quick reading
    def art_band(value):
        if value < 50:
            return "Low"
        elif value < 65:
            return "Medium"
        else:
            return "High"

    excel_df["ART_Coverage_Band"] = excel_df["ART_Coverage_Percent"].apply(art_band)

    display(excel_df)

except Exception as e:
    print("Error while transforming data:", e)
```

### 5.2 Export to Excel with Conditional Formatting and a Chart
`openpyxl` is used to write the transformed table to a `.xlsx` workbook, apply a colour-scale conditional format to the two key indicator columns, and embed a native Excel line chart.

```python
try:
    from openpyxl import Workbook
    from openpyxl.styles import Font, PatternFill, Alignment
    from openpyxl.formatting.rule import ColorScaleRule
    from openpyxl.chart import LineChart, Reference
    from openpyxl.utils import get_column_letter

    EXCEL_PATH = "SA_Health_Analysis.xlsx"

    wb = Workbook()
    ws = wb.active
    ws.title = "SA Health Data"

    headers = list(excel_df.columns)
    ws.append(headers)

    # Header styling
    for cell in ws[1]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = PatternFill(start_color="1F4E78", end_color="1F4E78", fill_type="solid")
        cell.alignment = Alignment(horizontal="center")

    # Write data rows
    for row in excel_df.itertuples(index=False):
        ws.append(list(row))

    # Auto-fit-ish column widths
    for i, col in enumerate(headers, start=1):
        ws.column_dimensions[get_column_letter(i)].width = max(18, len(col) + 2)

    n_rows = len(excel_df) + 1

    # Conditional formatting: ART coverage (green = high coverage = good)
    art_col_letter = get_column_letter(headers.index("ART_Coverage_Percent") + 1)
    art_rule = ColorScaleRule(
        start_type="min", start_color="F8696B",
        mid_type="percentile", mid_value=50, mid_color="FFEB84",
        end_type="max", end_color="63BE7B"
    )
    ws.conditional_formatting.add(f"{art_col_letter}2:{art_col_letter}{n_rows}", art_rule)

    # Conditional formatting: mortality (green = low mortality = good, reversed scale)
    mort_col_letter = get_column_letter(headers.index("Under5_Mortality_per_1000") + 1)
    mort_rule = ColorScaleRule(
        start_type="min", start_color="63BE7B",
        mid_type="percentile", mid_value=50, mid_color="FFEB84",
        end_type="max", end_color="F8696B"
    )
    ws.conditional_formatting.add(f"{mort_col_letter}2:{mort_col_letter}{n_rows}", mort_rule)

    # Native Excel line chart: ART coverage vs mortality over time
    chart = LineChart()
    chart.title = "ART Coverage vs Under-5 Mortality (South Africa, 2010-2024)"
    chart.x_axis.title = "Year"
    chart.y_axis.title = "Value"

    year_col = headers.index("Year") + 1
    art_col = headers.index("ART_Coverage_Percent") + 1
    mort_col = headers.index("Under5_Mortality_per_1000") + 1

    art_data = Reference(ws, min_col=art_col, min_row=1, max_row=n_rows)
    mort_data = Reference(ws, min_col=mort_col, min_row=1, max_row=n_rows)
    categories = Reference(ws, min_col=year_col, min_row=2, max_row=n_rows)

    chart.add_data(art_data, titles_from_data=True)
    chart.add_data(mort_data, titles_from_data=True)
    chart.set_categories(categories)
    chart.width, chart.height = 22, 11
    ws.add_chart(chart, f"A{n_rows + 3}")

    wb.save(EXCEL_PATH)
    print(f"Excel workbook saved to '{EXCEL_PATH}' with conditional formatting and a chart.")

except Exception as e:
    print("Error while creating Excel workbook:", e)
```
